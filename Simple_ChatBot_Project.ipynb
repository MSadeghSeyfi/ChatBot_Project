```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

from langchain_community.document_loaders import PyPDFium2Loader

pdf_path = "data/justforfun_persian.pdf"
loader = PyPDFium2Loader(pdf_path)
pdf_docs = loader.load()
```

```python
def replace_white_spaces(text):
    text = text.replace(chr(876) + ' ', 'ی ')
    text = text.replace(chr(876) + chr(13), 'ی ')
    text = text.replace('ک', chr(887))  # ͷ = chr(887) = U+0377
    return text

cleaned_docs = [replace_white_spaces(docs.page_content) for docs in pdf_docs] 
```
